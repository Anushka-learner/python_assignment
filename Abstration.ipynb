# Abstration

```python
from abc import ABC, abstractmethod
```

```python
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("Cat meows")

d1=Dog()
c1=Cat()
d1.sound()
c1.sound()
```
