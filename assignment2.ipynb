```python
numbers=[10, 15, 20, 25, 30]
new=["even" if i%2==0 else "odd" for i in numbers]
print(new)
```

```python
numbers=[10, -5, 20, -8, 30, -2]
new=[0 if i<0 else i for i in numbers]
print(new)
```

```python
prices=[500, 1200, 800, 2000, 1500]
new=[i*90/100 if i>=1000 else i*95/100 for i in prices]
print(new)
```

```python
employees=[
    {"name": "Amit","department": "IT"},
    {"name": "Rahul","department": "HR"},
    {"name": "Neha","department": "IT"},
    {"name": "Rohit","department": "Sales"}
    ]
new=[i["name"] for i in employees if i["department"]=="IT"]
print(new)
```
