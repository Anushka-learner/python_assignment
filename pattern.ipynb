```python
n=5
for i in range(1, 2*n+1):
    if i>n:
        space=i-n
        star=2*n-i
    else:
        space=n-i
        star=i
    for j in range(space):
        print("", end=" ")
    for k in range(star):
        print("*", end=" ")
    print(end="\n")
```
