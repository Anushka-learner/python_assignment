# Public

```python
class bankAccount:
    def __init__(self, name, salary):
        self.name=name
        self.salary=salary     #public
    def showdetails(self):
        print("Name: ", self.name)
        print("Salary: ", self.salary)

e1=bankAccount("Avnii", 65000)
e1.showdetails()
print("="*25)
print(e1.salary)
```

# Protected

```python
class employee:
    def __init__(self, name, salary):
        self.name=name
        self._salary=salary    #protected
    def showdetails(self):
        print("Name: ", self.name)
        print("Salary: ", self._salary)

e1=employee("Avnii", 65000)
e1.showdetails()
print("="*25)
print(e1._salary)
```

# Private

```python
class employee:
    def __init__(self, name, salary):
        self.name=name
        self.__salary=salary    #private
    def showdetails(self):
        print("Name: ", self.name)
        print("Salary: ", self.__salary)

e1=employee("Avnii", 65000)
e1.showdetails()
print("="*25)
print(e1.__salary)
```

```python
class BankAccount():
    def __init__(self, name, balance, pin):
        self.__name=name  # Set
        self.__balance=balance  # Set
        self.__pin=pin  # Set
    def setName(self, name):
        self.__name=name  # Set
        
    def setBalance(self, balance):
        self.__balance=balance  # Set
        
    def setPin(self, pin):
        self.__pin=pin  # Set
        
    def showAccDetails(self):
        print("Name: ", self.__name)  # Get
        print("Balance: ", self.__balance)  # Get
        print("User Pin: ", self.__pin)  # Get

bank=BankAccount("Shreya", 2300, "shr123")
bank.showAccDetails()
print("="*25)

bank.setName("Shagun")
bank.showAccDetails()
print("="*25)

bank.setPin("shg123")
bank.showAccDetails()
print("="*25)
```

```python
#Complete Encapsulation Example
class BankAccount():
    def __init__(self, name, balance, pin):
        self.__name=name  # Set
        self.__balance=balance  # Set
        self.__pin=pin  # Set
        
    def setName(self, name):
        self.__name=name  # Setter
        
    def setBalance(self, balance):
        self.__balance=balance  # Setter
        
    def setPin(self, pin):
        self.__pin=pin  # Setter

    def getName(self):
        return self.__name  # Getter
        
    def getBalance(self):
        return self.__balance  # Getter
        
    def getPin(self):
        return self.__pin  # Getter
        
    def showAccDetails(self):
        print("Name: ", self.__name)  # Get
        print("Balance: ", self.__balance)  # Get
        print("User Pin: ", self.__pin)  # Get

obj=BankAccount("Shreya", 2300, "shr123")
print(obj.getName())
print(obj.getBalance())
print(obj.getPin())
print("="*30)
```
