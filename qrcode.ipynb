```python
import qrcode
url=input("Enter URL to make QR: ")

if url.startswith("http://") or url.startswith("https://"):
    qr=qrcode.make(url)
    qr.save("abcd.png")
    print("Ban gaya")
else:
    print("Invalid URL... Must start with http or https")
```

```python
import qrcode
url=input("Enter URL to make QR: ")

if url.startswith("http://") or url.startswith("https://"):
    qr=qrcode.QRCode(version=1,
                    error_correction=qrcode.constants.ERROR_CORRECT_H,
                    box_size=10,
                    border=4
                    )
    qr.add_data(url)
    qr.make()
    image=qr.make_image(fill_color="black", back_color="pink")
    image.save("FirstQR.png")
else:
    print("Invalid URL")
```

```python
#create QR Code For UPI
import qrcode

upi_id=input("Enter UPI ID: ")
name=input("Enter Name: ")
upi_url=f"upi://pay?pa={upi_id}&pn={name}"

#pa=payee
#pn=payee name
qr=qrcode.make(upi_url)
qr.save("Anu_UPI_1.png")
print("UPI ka QR Ban Gaya")
```

```python
#create QR Code For UPI with specific amount
import qrcode

upi_id=input("Enter UPI ID: ")
name=input("Enter Name: ")
amount=int(input("Enter Amount: "))
upi_url=f"upi://pay?pa={upi_id}&pn={name}&am={amount}"

#pa=payee
#pn=payee name
#am=amount
#cu=INR
qr=qrcode.make(upi_url)
qr.save("Anu_UPI_1.png")
print("UPI ka QR Ban Gaya")
```
